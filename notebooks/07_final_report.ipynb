### 3. Key Findings


#### 3.1 Daily temperature association
The audit recomputation on the tracked raw CSVs included **565,118 complaint rows** in 2024 and **366 matched days**. Pearson correlation between daily complaint count and the midpoint of daily maximum/minimum temperature is **r = 0.5803**. The previous prose value, 0.12, conflicted with the saved output and has been corrected.

Run `python scripts/validate_daily_correlation.py` from the repository root to reproduce this aggregate check. The raw numeric weather scale is preserved; temperature is a midpoint proxy rather than a measured daily mean.

The nominal Pearson p-value assumes independent observations. Seasonality and temporal autocorrelation have not been adjusted, so it should not be presented as a validated causal or inferential conclusion.


```python
# Example code to compute Pearson correlation
import pandas as pd
from scipy.stats import pearsonr

crime = pd.read_csv('../data/processed/crime_clean.csv', parse_dates=['date'])
weather = pd.read_csv('../data/processed/weather_clean.csv', parse_dates=['date'])
weather['temp_avg'] = (weather['temp_max'] + weather['temp_min']) / 2
merged = pd.merge(
    crime.groupby('date').size().reset_index(name='crime_count'),
    weather[['date','temp_avg']], on='date'
)
r, p = pearsonr(merged['crime_count'], merged['temp_avg'])
print(f"Pearson r={r:.2f}, p-value={p:.3f}")
```

### 4. Conclusions
The tracked 2024 snapshot shows a positive descriptive association between daily recorded complaint counts and temperature. This is compatible with multiple explanations, including shared seasonality; no causal effect has been identified.

This analysis does not establish that heat causes crime, that precipitation changes violent crime, or that additional patrols would improve outcomes. Operational policing recommendations have therefore been removed.

A stronger study would specify an estimand, address temporal dependence and seasonality, examine reporting/measurement biases and evaluate interventions separately.


### 5. Limitations and future work

- The available snapshot covers one year and one weather station.
- Spatial and offense-specific patterns require their own validation.
